```python
print("Welcome to Traseure Island")
print("Your mission is to find the treasure.")
```

```python
crossroad = input(" You're at a crossroad. Where do you want to go? Type 'left' or 'right'.").lower()

if crossroad == "left":
  lake = input("You've come to a lake. There is an island in the middle of the lake. Type 'wait' to wait for a boat. Type 'swim' to swim across.").lower()
  if lake == "wait":
    door = input("You arrive at the island unharmed. There is a house with 3 doors. One red, one yellow and one blue. Which colour do you choose?").lower()
    if door == "yellow":
      print("You found the treasure! You Win!")
    elif door == "red":
      print("It's a room full of fire. Game Over.")
    elif door == "blue":
      print("You enter a room of beasts. Game Over.")

else:
  print("Game over")

```
